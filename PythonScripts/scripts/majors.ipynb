## Chuẩn hóa danh sách và chuỗi

```python
import re

def remove_newlines(val):
    if pd.isna(val):
        return val
    if isinstance(val, str):
        return re.sub(r'[\r\n]+', ' ', val).strip()
    return val

df = df.map(remove_newlines)

def normalize_score_codes(val):
    if pd.isna(val) or str(val).strip() == '':
        return val
    s = str(val).replace(',', ';')
    s = re.sub(r'\s+', '', s)
    return s

df['score_codes'] = df['score_codes'].apply(normalize_score_codes)
```

# Lưu

```python
output_path = r'D:\UNIDOCS\OLAP\IS217.R11-OLAP\PythonScripts\cleaned-data\cleaned-majors.csv'
df.to_csv(output_path, index=False)
df.info()
df.head()
```
