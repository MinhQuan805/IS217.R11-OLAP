# Chuẩn hóa dữ liệu

```python
for col in df.columns:
    print(f"\n=== Cột: {col} ===")
    print(f"Số giá trị độc nhất: {df[col].nunique()}")
    print(df[col].value_counts(dropna=False))
```

Giải pháp: Chuẩn hóa lại dữ liệu đối với subject1, subject2, subject3:
- Chỉ in hoa chữ cái đầu, các chữ cái còn lại in thường.
- Loại bỏ khoảng trắng ở đầu và cuối chữ và không có 2 dấu cách liên tiếp.
- Thay thế `lý` thành `lí`.

```python
import re

def normalize_text(text):
    if pd.isna(text):
        return text
    
    # 1. Loại bỏ khoảng trắng đầu/cuối
    text = text.strip()
    
    # 2. Loại bỏ 2 dấu cách liên tiếp trở lên -> thành 1 dấu cách
    text = re.sub(r'\s+', ' ', text)
    
    # 3. Đưa về chữ thường toàn bộ trước
    text = text.lower()
    
    # 4. Thay thế "lý" thành "lí"
    text = text.replace('lý', 'lí')
    
    # 5. In hoa chữ cái đầu tiên, còn lại giữ nguyên (chữ thường)
    if text:
        text = text[0].upper() + text[1:]
    
    return text

# Áp dụng cho 3 cột
for col in ['subject1', 'subject2', 'subject3']:
    df[col] = df[col].apply(normalize_text)

# Kiểm tra lại kết quả
for col in ['subject1', 'subject2', 'subject3']:
    print(f"\n=== Cột: {col} ===")
    print(df[col].value_counts(dropna=False))
```
